```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier

# 1. Load the data
churn= pd.read_csv (r"C:\Users\USER\Downloads\churn.csv")
# 2. Preprocessing
le = LabelEncoder()
churn['International plan'] = le.fit_transform(churn['International plan'])
churn['Voice mail plan'] = le.fit_transform(churn['Voice mail plan'])
churn['Churn'] = churn['Churn'].astype(int)

# 3. Define Features and Target
# We drop 'State' and 'Churn' to create the feature set X
X = churn.drop(['Churn', 'State'], axis=1)
y = churn['Churn']

# 4. Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Define and Train the Model (best_rf)
# Using the best parameters from our Grid Search
best_rf = RandomForestClassifier(max_depth=10, n_estimators=200, random_state=42)
best_rf.fit(X_scaled, y)

# 6. Add predictions to the dataframe
churn['Predicted_Churn'] = best_rf.predict(X_scaled)

# 7. Save the final file for Power BI
churn.to_csv('churn_with_predictions.csv', index=False)

print("Process Complete! Your file 'churn_with_predictions.csv' is ready for Task 2.")
```
